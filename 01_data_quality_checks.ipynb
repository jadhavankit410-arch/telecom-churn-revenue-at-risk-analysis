# 01 - Data Quality Checks
Load the CSV, clean column names, check nulls, duplicates, customer status and the Total Revenue formula.

```python
import pandas as pd

df = pd.read_csv("../01_data/raw/telecom_customer_churn.csv")
print(df)
print(df.shape)
```

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
print(df.columns.tolist())
```

```python
print(df.dtypes)
print()
nulls = df.isnull().sum()
print(nulls[nulls > 0])
```

```python
print("Duplicate customer IDs:", df["customer_id"].duplicated().sum())
print()
print(df["customer_status"].value_counts())
print()
print(df["customer_status"].value_counts(normalize=True).round(3))
```

```python
calc = (df["total_charges"] + df["total_extra_data_charges"]
        + df["total_long_distance_charges"] - df["total_refunds"]).round(2)
print("Revenue mismatches:", (calc != df["total_revenue"].round(2)).sum())
```
